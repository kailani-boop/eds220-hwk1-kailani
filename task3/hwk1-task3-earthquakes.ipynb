# 6
Without creating any new variables, plot a histogram of the eartquake's magnitudes (`eqk` data). Add a title and change the color of your graph.

```python
# Import package to plot things
import matplotlib.pyplot as plt

# Plotting the magnitude of earthquakes
# Magnitude is on the x-axis, the count is on the y-axis. Bar color and edge color were changed
plt.hist(data = eqk, x = 'mag', color = 'lightblue', edgecolor = 'black')

# Adding a title
plt.title('Histogram of Earthquakes')

# Adding a x and y intercept label
plt.xlabel('Magnitude')
plt.ylabel('Count')
```

<!-- END QUESTION -->

# 7

a. Create a new variable `top20` with the 20 earthquakes with largest magnitude. HINT: check the method [`pandas.Series.nlargest`](https://pandas.pydata.org/docs/reference/api/pandas.Series.nlargest.html). Your answer should contain only the `id` and `mag` columns. 

```python
# New variable of the top 20 earthquakes with the largest magnitude
# Only keeps the index (id) and the 'mag' column
top20 = eqk.nlargest(20, columns = 'mag')[['mag']]
```

```python
# View the dataframe 
top20
```

```python
grader.check("q7_a")
```

<!-- BEGIN QUESTION -->

b. Create a bar plot showing the magnitude of the top 20 earthquakes. Update the title, x-axis label, y-axis label, and color of the bars. 

```python
# Making a barplot of the top 20 earthquakes
# X is now the ID of the earthquake, which is also now the index of eqk dataframe
# Height is the magnitude, color was changed
plt.bar(data = top20, x = top20.index, height = 'mag', color = 'darkred', edgecolor = 'black')

# Adding a title
plt.title('Barplot of Top 20 Earthquakes')

# Adding a x and y intercept label
plt.xlabel('ID')
plt.ylabel('Magnitude')

# Rotating the ID's so they're readable
plt.tick_params(axis='x', labelrotation=90)
```

# 9 

Without creating any new variables, visualize the locations of eartquakes with magnitude greter than 5 by creating a scatter plot of their latitude and longitude. Make the following adjustments to the basic plot:
- update the title
- color the points by magnitude (HINT: [check the `c` parameter here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.plot.scatter.html))
- adjust the `alpha` (transparency) argument
- adjust the `colormap` argument to maket it look prettier. [Click for ideas of sequential colormaps.](https://matplotlib.org/stable/users/explain/colors/colormaps.html)


```python
# Filtering the magnitudes to be over 5 without creating a new variable
# Making a scatter plot with x as the latitude and y as the longitude
# The points are where earthquakes occurred (c = 'mag'). Alpha changes the transparency. 
# Colormap gives a color spectrum
eqk[eqk['mag'] > 5].plot.scatter(x = 'latitude', y = 'longitude', c = 'mag', alpha = 1.0, colormap = 'Reds')

# Adding a title
plt.title('Scatterplot of Earthquakes with Magnitudes > 5')

# Adding a x and y intercept label
plt.xlabel('Latitude')
plt.ylabel('Longitude')
```
